# 🔵 **1. Association Analysis untuk Atribut Kategorikal**

## 📘 Permasalahan

Algoritma Apriori/FP-Growth standar hanya bekerja untuk **binary attributes** → sebuah atribut hanya bernilai hadir (1) atau tidak hadir (0).

Contoh data kategorikal seperti:

- Browser \= IE / Mozilla / Netscape
- Country \= US, China, Germany, Etc. (200 kemungkinan nilai)
- Buy \= Yes / No

tidak bisa langsung digunakan.


````python
df = pd.DataFrame({
    'country': ['USA','China','USA','Germany','Australia','India','Brazil','USA','India','France'],
    'browser': ['IE','Netscape','Mozilla','IE','Mozilla','Chrome','Chrome','Mozilla','IE','Mozilla'],
    'buy': ['No','No','Yes','Yes','No','No','Yes','No','No','Yes']
})
print("Sample data:")
df

````


## 🎯 Solusi: *Itemization (One-hot Encoding)*

Setiap kombinasi atribut-nilai diubah menjadi **item baru**.

Contoh:
Browser memiliki 3 nilai → menjadi:

- Browser\=IE
- Browser\=Netscape
- Browser\=Mozilla

## 🔍 Tantangan

1. **Item bisa sangat banyak**
    Jika atribut kategorikal memiliki puluhan nilai.
2. **Nilai langka** (rare categories)
    Support rendah → sering hilang.
3. **Nilai dominan**
    Misal "Buy\=No" muncul 95% → rule tidak menarik.

````python
# One-hot (asymmetric binary representation)
ohe = pd.get_dummies(df[['country','browser']], prefix=['country','browser'])
print("\nOne-hot sample:")
ohe.head()
````

````python
# Aggregate low-support categories into 'Other'
counts = df['country'].value_counts()
low_support_vals = counts[counts < 2].index.tolist()   # adjust threshold as needed
df['country_grouped'] = df['country'].apply(lambda x: x if x not in low_support_vals else 'Other')
print("\nGrouped countries:")
print(df[['country','country_grouped']])
````

````python
# Consider dropping attributes with extremely skewed values
print("\nBuy distribution (fraction):")
print(df['buy'].value_counts(normalize=True))
````

````python
from mlxtend.frequent_patterns import apriori

frequent_itemsets = apriori(ohe, min_support=0.1, use_colnames=True)

print(frequent_itemsets)
````

````python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.6)
ruleDf = pd.DataFrame(rules[['antecedents','consequents', 'support','confidence','lift']])
ruleDf
````

# 🔵 **2. Association Analysis untuk Atribut Kontinu**

## 📘 Permasalahan

Atribut numerik seperti:

- Salary
- Age
- Duration
- Session length

tidak bisa langsung dipakai sebagai item.

````python
rng = np.random.default_rng(42)
df_cont = pd.DataFrame({
    'age': rng.integers(15,80,size=100),
    'income': np.round(rng.normal(50000,15000,size=100),0),
    'buy': rng.choice(['Yes','No'], size=100, p=[0.25,0.75])
})
df_cont
````

Catatan: pilih metode binning berdasarkan kebutuhan — equal-frequency menjaga jumlah dalam tiap bin (stabil support).

## 🎯 Solusi: Discretization / Binning

Mengubah nilai numerik menjadi interval, misalnya:

```
Age = 22 → Age=[20–30)
Age = 45 → Age=[40–50)
```

### Dua metode utama:

1. **Equal-width binning**
    Interval dengan lebar sama
2. **Equal-depth (quantile) binning**
    Setiap interval memiliki jumlah data yang sama

### Masalah penting:

- Interval terlalu lebar → rule tidak menarik
- Interval terlalu sempit → support terlalu rendah
- Mencoba semua interval → kompleksitas O(n²)

````python
# Equal-width binning
df_cont['age_bin_eqw'] = pd.cut(df_cont['age'], bins=4, labels=False)

# Equal-frequency binning (quantile)
df_cont['age_bin_eqf'] = pd.qcut(df_cont['age'], q=4, labels=False, duplicates='drop')

df_cont[['age','age_bin_eqw','age_bin_eqf']].head()
````

````python
# Show fraction (support) per bin
print("\nSupport per equal-width bin:")
df_cont['age_bin_eqw'].value_counts(normalize=True).sort_index()
````

````python
print("\nSupport per equal-frequency bin:")
df_cont['age_bin_eqf'].value_counts(normalize=True).sort_index()
````

````python
# Create labels for association mining
df_cont['age_label'] = df_cont['age_bin_eqf'].apply(lambda x: f'age_q{x}')
df_cont['income_label'] = pd.qcut(df_cont['income'], 4, labels=['inc_q0','inc_q1','inc_q2','inc_q3'], duplicates='drop')
transactions = df_cont[['age_label','income_label','buy']].values.tolist()
print("\nExample transactions (first 5):",)

df = pd.DataFrame({'transaction': transactions})
df.head()
````

# 🔵 **3. Statistical-Based Association Analysis**


Ketika asosiasi bukan item biner, tetapi **angka**, misalnya:

```
Browser=Mozilla AND Buy=Yes → Age = 23
```

## 📘 Langkah:

1. Pilih subset data yang memenuhi rule
2. Hitung statistik (mean, variance)
3. Bandingkan dengan data yang tidak memenuhi rule
4. Gunakan uji statistik (Z-test) untuk cek signifikansi

## 🎯 Tujuan

Menentukan apakah rule tersebut:
- kebetulan
- atau benar-benar signifikan secara statistik


````python
# Contoh antecedent: high income & buy=Yes → consequent: rata-rata umur (age)
ante_mask = (df_cont['income'] > df_cont['income'].median()) & (df_cont['buy'] == 'Yes')
group_mean = df_cont.loc[ante_mask, 'age'].mean()
group_n = ante_mask.sum()
group_std = df_cont.loc[ante_mask, 'age'].std(ddof=1)

comp_mean = df_cont.loc[~ante_mask, 'age'].mean()
comp_n = (~ante_mask).sum()
comp_std = df_cont.loc[~ante_mask, 'age'].std(ddof=1)

print(f'Antecedent mean age = {group_mean:.2f} (n={group_n}), Complement mean = {comp_mean:.2f} (n={comp_n})')
````

````python
# Uji hipotesis sederhana (one-sided): H0: mu_group <= mu_comp  vs H1: mu_group > mu_comp
delta = 0.0
num = (group_mean - comp_mean) - delta
den = np.sqrt((group_std**2)/group_n + (comp_std**2)/comp_n)
z = num/den if den > 0 else np.nan

from scipy import stats
pval_one_sided = 1 - stats.norm.cdf(z)  # jika z besar positif -> p kecil
print("z-statistic:", z, "one-sided p-value:", pval_one_sided)
````



## 📌 Interpretasi

Jika Z-score besar dan p-value kecil → rule signifikan.
Ini perlu diperhatikan ketika kita ingin membangun **numeric prediction rules**.
Penjelasan: jika p-value kecil (mis. <0.05), kita bisa menyatakan segmen yang dimodelkan antecedent memiliki mean yang berbeda secara signifikan.

# 🔵 **4. Min-Apriori (for Freq. Count Data)**

Min-Apriori diperkenalkan ketika dataset berbentuk:

- Document-term matrix
- word frequencies
- continuous counts

````python
df = pd.DataFrame({
    'W1':[2,0,2,0,1],
    'W2':[2,0,3,0,1],
    'W3':[0,1,0,1,1],
    'W4':[0,2,0,0,0],
    'W5':[1,2,0,1,2]
}, index=['D1','D2','D3','D4','D5'])
df
````

## 📘 Permasalahan

Jika menghitung support secara langsung → bisa \> jumlah dokumen → salah.

## 🎯 Solusi: Normalisasi per dokumen

Gunakan L1-norm:

```
freq / total_freq
```

Sehingga setiap baris \= 1.

## 📘 Support min-Apriori

$$
sup(X) = \sum_i \min_j D(i,j)
$$

D(i,j) \= nilai normalisasi dokumen i untuk item j.

````python
# Normalisasi L1 per dokumen (baris)
D_norm = df.div(df.sum(axis=1), axis=0).fillna(0)
print("Normalized document-term matrix:")
print(D_norm.round(3))
````

````python
# fungsi min-apriori support: sum per dokumen nilai minimum untuk itemset
def min_apriori_support(itemset, D_norm):
    return D_norm[list(itemset)].min(axis=1).sum()

print("\nMin-Apriori support for {W1,W2,W3}:", min_apriori_support(['W1','W2','W3'], D_norm))
````

## 📌 Interpretasi

Anti-monotone property tetap berlaku.
Itemset lebih besar → support menurun.

Penjelasan: Min-Apriori menghitung kontribusi parsial dari dokumen (nilai antara 0..1) dan menjumlahkannya sehingga support berskala kontinu tepat untuk data frekuensi.

# 🔵 5. Sequence Data & Sequintial Pattern Mining

**sequence** didefinisikan sebagai:

$$
s = \langle e_1, e_2, e_3, ...\rangle
$$

di mana setiap elemen **eᵢ** adalah *itemset* (set item yang terjadi bersama pada waktu tertentu).

Contoh:

```
< {a b} {c} {d e} >
```

Interpretasi:

- `{a b}` terjadi bersama pada waktu t1
- `{c}` terjadi di t2
- `{d e}` pada t3
- Ini adalah data *urutan event*.

Sequence mining bertujuan mencari **pola yang sering muncul dalam urutan**.

**Subsequence** adalah urutan yang menjaga **order**, tapi tidak harus berdekatan.

Contoh sequence:

```
<A B> <C> <D F>
```

Subsequence valid:

- `<A>`
- `<A> <C>`
- `<A> <D>`
- `<B> <F>`
- `<A B> <F>`

Subsequence *tidak valid* (karena melanggar urutan):

- `<C> <A>`
- `<F> <D>`

#### **Support dalam Sequential Pattern Mining**

Support adalah:

$$
\text{support}(P) = \frac{\text{jumlah sequence yang mengandung P}}{\text{total sequence}}
$$

Mirip Apriori, tapi yang dicek adalah **apakah sequence mengandung subsequence**.

##### **Tantangan Utama**

Jika ada 9 item dalam sequence:

```
<{a b} {c d e} {f} {g h i}>
```

Jumlah possible subsequences panjang 4:

$$
C(9,4) = 126 → Eksplosif.
$$

**Sequential mining jauh lebih berat** daripada Apriori.

**Contoh:**

````python
sequences = [
    [[1, 2], [2, 3], [5]],          # S1
    [[1, 2], [2, 3, 4]],            # S2
    [[1, 2], [2, 3, 4], [2, 4, 5]], # S3
    [[2], [3, 4], [4, 5]],          # S4
    [[1, 3], [2, 4, 5]]             # S5
]
sequences
````

##### Cek apakah subsequence muncul dalam sequence & Hitung Support Sequence

````python
def contains(seq, pattern):
    i = 0
    for elem in seq:
        if set(pattern[i]).issubset(set(elem)):
            i += 1
            if i == len(pattern):
                return True
    return False

def support(pattern, sequences):
    count = sum(contains(s, pattern) for s in sequences)
    return count / len(sequences)

print("Support <1> →", support([[1]], sequences))
print("Support <1><2> →", support([[1],[2]], sequences))
````

#### Menghitung Frequent 1-sequence (L1)

````python
items = sorted(set(item for seq in sequences for elem in seq for item in elem))
print("Items =", items)

minsup = 0.4

# Frequent 1-sequence
L1 = []
for it in items:
    pattern = [[it]]
    if support(pattern, sequences) >= minsup:
        L1.append(pattern)

print("Frequent 1-sequence (L1):", L1)
````

##### generasi Candidate 2-sequences (C2)
Kandidat 2-sequence harus mencakup:
- Item yang muncul bersama:

`<{i,j}>`

- Item berurutan:

`<{i}{j}>`

````python
import itertools
C2 = []

# case 1: same element (i,j) like {i,j}
for a, b in itertools.combinations(items, 2):
    C2.append([[a, b]])

# case 2: sequential <i><j>
for a in items:
    for b in items:
        C2.append([[a], [b]])

print("Total C2 candidates:", len(C2))
print("C2", C2)
````

##### Menghitung Frequent 2-sequence (L2)

````python
L2 = [c for c in C2 if support(c, sequences) >= minsup]

print("Frequent 2-sequences (L2):")

rows = []

for s in L2:
    rows.append({
        "sequence": s,
        "Support": support(s, sequences)
    })

df = pd.DataFrame(rows)
df
````

#### Menambahkan Max-Gap Constraint

Contoh: max-gap = 1

→ pola harus muncul dalam setidaknya ≤1 langkah waktu

````python
def contains_with_maxgap(seq, pattern, max_gap):
    i = 0
    last_pos = -1

    for pos, elem in enumerate(seq):
        if set(pattern[i]).issubset(set(elem)):
            # If last element was matched and gap is too large: fail
            if last_pos != -1 and (pos - last_pos) > max_gap:
                return False
            last_pos = pos
            i += 1
            if i == len(pattern):
                return True
    return False

print("Support <1><2> with max-gap=1:",
      sum(contains_with_maxgap(s, [[1],[2]], 1) for s in sequences) / len(sequences))
````
